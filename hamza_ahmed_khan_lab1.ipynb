# **✈️ Simple AI Travel Agent**

```python
print("Welcome to the Simple AI Travel Agent!")
print("Supported cities: Karachi, Islamabad, Lahore, Peshawar")
print("-" * 55)

departure = input("Where are you traveling from? ").strip().lower()
destination = input("Where would you like to travel to? ").strip().lower()

if departure == "karachi" and destination == "islamabad":
    airline, dep_time, arr_time, price = "PIA", "07:30 AM", "09:30 AM", "PKR 15,000"
elif departure == "karachi" and destination == "lahore":
    airline, dep_time, arr_time, price = "Airblue", "08:00 AM", "09:45 AM", "PKR 13,500"
elif departure == "karachi" and destination == "peshawar":
    airline, dep_time, arr_time, price = "Serene Air", "10:00 AM", "12:15 PM", "PKR 17,000"
elif departure == "islamabad" and destination == "karachi":
    airline, dep_time, arr_time, price = "PIA", "11:00 AM", "01:00 PM", "PKR 15,000"
elif departure == "islamabad" and destination == "lahore":
    airline, dep_time, arr_time, price = "Airblue", "06:45 AM", "07:45 AM", "PKR 9,000"
elif departure == "lahore" and destination == "karachi":
    airline, dep_time, arr_time, price = "Airblue", "05:30 PM", "07:15 PM", "PKR 13,500"
elif departure == "lahore" and destination == "peshawar":
    airline, dep_time, arr_time, price = "Serene Air", "03:30 PM", "04:45 PM", "PKR 11,000"
elif departure == "peshawar" and destination == "islamabad":
    airline, dep_time, arr_time, price = "Airblue", "08:30 AM", "09:20 AM", "PKR 8,500"
elif departure == "peshawar" and destination == "karachi":
    airline, dep_time, arr_time, price = "Serene Air", "04:00 PM", "06:15 PM", "PKR 17,000"
else:
    airline = None

if airline is None:
    print(f"\nSorry! Flights from {departure.title()} to {destination.title()} are not available.")
else:
    print("\nHere is the available flight:")
    print("-" * 35)
    print(f"From             : {departure.title()}")
    print(f"To               : {destination.title()}")
    print(f"Airline          : {airline}")
    print(f"Departure Time   : Tomorrow, {dep_time}")
    print(f"Arrival Time     : Tomorrow, {arr_time}")
    print(f"Ticket Price     : {price}")
    print("-" * 35)

    choice = input("Would you like to book this flight? (yes/no) ").strip().lower()

    if choice == "yes" or choice == "book the flight":
        print("\nYour flight has been booked successfully!")
    elif choice == "no":
        print("\nBooking cancelled.")
    else:
        print("\nNot understood. Booking cancelled.")
```
